```python
#Импорт библиотеки NumPY
import numpy as np

#Примеры создания массивов

#Создание массива тз обычного списка
arr_list = np.array([1, 2, 3, 4, 5])
print("Исходный массив:", arr_list)

#Создание массива заполненного нулями
arr_zeros = np.zeros((2,3))
print("Массив нулей:", arr_zeros)

#Создание массива заданного единицами
arr_ones = np.ones((3,2))
print("Массив единиц:", arr_ones)

#Создание массива с помощью arange
arr_arange = np.arange(0, 8, 2)
print("Массив с помощью arange", arr_arange)

#Демонстрация атрибутов

#Форма массива
arr_shape = arr_ones.shape
print("Форма массива единиц:", arr_shape)
#Тип данных массива
arr_dtype = arr_ones.dtype
print("Тип данных массива единиц:", arr_dtype)

#Индексирование и среды
# Получение отдельного элемента
single_element = arr_list[2]  # третий элемент массива
print("Отдельный элемент:", single_element)

# Извлечение подмассива с помощью среза
subarray = arr_list[1:4]  # элементы со 2-го по 4-й
print("Срез массива:", subarray)

# --- Векторные операции и статистика ---

# Поэлементное сложение двух массивов
arr_sum = arr_list + np.array([5, 4, 3, 2, 1])
print("Поэлементная сумма:", arr_sum)

# Поэлементное умножение
arr_mul = arr_list * 2
print("Поэлементное умножение:", arr_mul)

# Вычисление средней величины
mean_value = arr_list.mean()
print("Среднее значение:", mean_value)
```
